```python
import pandas as pd
df = pd.read_csv(r'C:\Documents\Data analytics\projects\my-data-project\E-COMMERCE-PROJECT\data\raw\Amazon Sale Report.csv')
```

```python
df_copy = df.copy()
df_copy.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')
```

```python
df_copy.head()  
```

```python
df_copy.drop(columns=['index', 'unnamed:_22'], inplace=True)
```

```python
df_copy.drop(columns=['ship_country', 'currency', 'ship_postal_code', 'promotion_ids', 'courier_status', 'asin'], inplace=True)
```

```python
string_cols = df_copy.select_dtypes(include=['object', 'string']).columns
for col in string_cols:
    df_copy[col] = (df_copy[col]
                    .fillna('')
                    .str.strip()
                    .str.lower()
                    .str.replace('-', '_', regex=False)
                    .str.replace(' ', '_', regex=False)
                    .replace('', pd.NA))
```

```python
correct_data_types = {
    'date': 'datetime64[ns]',
    'status': 'string',
    'fulfilment': 'string',
    'sales_channel': 'string',
    'ship_service_level': 'string',
    'style': 'string',
    'sku': 'string',
    'category': 'string',
    'size': 'string',
    'qty': 'int',
    'amount': 'float',
    'ship_city': 'string',
    'ship_state': 'string',
    'b2b': 'bool',
    'fulfilled_by': 'string',
}
df_copy = df_copy.astype(correct_data_types)
df_copy.dtypes

```

```python
df_copy.sample(20)
```

```python
df_copy.dtypes
```

```python
#drop 
```

```python
df_copy.drop(columns=['order_id'], inplace=True)
```

```python
import numpy as np
missingvals = ['N/A', 'None', 'none', 'NONE', 'n/a', '?', 'nan', 'na', 'null', 'Null', 'NULL', '<NA>', 'NA']
df_copy = df_copy.replace(missingvals, np.nan)
```

```python
df_copy.sample(10)
```

```python
df_cleaned = df_copy.drop_duplicates()
```

```python
df_cleaned['status'].unique()
```

```python
df_cleaned['status'] = df_cleaned['status'].str.replace('___', '_')
```

```python
df_cleaned.sample(20)
```

```python
df_cleaned['ship_state'].unique()
```

```python
state_mapping = {
    'orissa' : 'odisha',
    'punjab/mohali/zirakpur' : 'punjab',
    'nl' : 'nagaland',
    'rajsthan' : 'rajasthan',
    'rajshthan' : 'rajasthan',
    'up' : 'uttar_pradesh',
    'pb' : 'punjab',
    'ar' : 'arunachal_pradesh',
    'rj' : 'rajasthan',
    'apo' : 'delhi',
    'new_delhi' : 'delhi',
    'chandigarh' : 'punjab',
    'ladakh' : 'jammu_&_kashmir',
    'puducherry' : 'tamil_nadu',
    'pondicherry' : 'tamil_nadu',
    

   
}
df_cleaned['ship_state'] = df_cleaned['ship_state'].replace(state_mapping)
```
